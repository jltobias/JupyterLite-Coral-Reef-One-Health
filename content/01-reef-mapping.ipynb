# Explore the reef mapping catalogue

**Goal:** compare the purposes and limitations of three mapping initiatives, then identify what to check before obtaining real data.

This notebook reads a bundled descriptive catalogue. It contains no satellite imagery or reef habitat polygons and needs no API keys or extra packages. Open the notebook in JupyterLite Lab, wait for Python, then choose **Run > Run All Cells**. Source links open provider websites separately.

```python
import json
from pathlib import Path
from html import escape
from IPython.display import HTML, display

data_path = Path("data/initiatives.json")
if not data_path.exists():
    raise FileNotFoundError("Open this notebook beside the data folder. In JupyterLite, refresh after the service worker starts and allow site storage.")
initiatives = json.loads(data_path.read_text(encoding="utf-8"))
print(f"Loaded {len(initiatives)} mapping initiative descriptions (not provider datasets).")
```

## Compare purpose and coverage

The descriptions are paraphrases of the linked sources. Product details and access terms can change; consult live metadata before using real data.

```python
headers = ["Initiative", "Approach", "Scope", "Caution"]
keys = ["initiative", "approach", "scope", "caution"]
rows = "".join("<tr>" + "".join(f"<td>{escape(item[key])}</td>" for key in keys) + "</tr>" for item in initiatives)
display(HTML("<table><thead><tr>" + "".join(f"<th>{heading}</th>" for heading in headers) + "</tr></thead><tbody>" + rows + "</tbody></table>"))
```

## Inspect one initiative

Change `selection` to `0`, `1`, or `2` and rerun the cell. What question could this resource help answer? What additional observations would you need?

```python
selection = 0
selected = initiatives[selection]
print(selected["initiative"])
print("Strength:", selected["strength"])
print("Caution:", selected["caution"])
display(HTML(f'<a href="{escape(selected["source_url"], quote=True)}" target="_blank" rel="noopener noreferrer">Visit the source</a>'))
```
